```python
def convertText2Ids(text):
    return tokenizer.encode_plus(
        text,
        max_length=512,
        add_special_tokens=True, # Add '[CLS]' and '[SEP]'
        return_token_type_ids=True,
        pad_to_max_length=True,
        return_attention_mask=True,
        truncation=True
    )
```

```python
# BertTokenizer = None
# BertTokenizer = bert.bert_tokenization.FullTokenizer
bert_layer = hub.KerasLayer("https://tfhub.dev/tensorflow/bert_en_uncased_L-12_H-768_A-12/1",
                            trainable=False)
vocabulary_file = bert_layer.resolved_object.vocab_file.asset_path.numpy()
to_lower_case = bert_layer.resolved_object.do_lower_case.numpy()
tokenizer = BertTokenizer(vocabulary_file, to_lower_case)
```

```python
# def textToToken(text):
#     #Train Data
#     return tokenizer(
#         text=text,
#         add_special_tokens=True,
#         max_length=200,
#         truncation=True,
#         padding=True, 
#         return_tensors='tf',
#         return_token_type_ids = False,
#         return_attention_mask = True,
#         verbose = True
#     )

def textToToken(text):
    #Train Data
    return tokenizer(
        text=text,
        add_special_tokens=True,
        max_length=200,
        truncation=True,
        padding='max_length', 
        return_tensors='tf',
        return_token_type_ids = False,
        return_attention_mask = True,
        verbose = True
    )
```

```python
# Model BERT1 Fix
print('Loading Model...')
modelBERT1 = tf.keras.models.load_model('Model/BERTmodel1.h5')
print('Loading Model Completed!')
queryBERT1 = str(input('Write some sentence: '))
queryBERT1 = textToToken(queryBERT1)
BERTPred = modelBERT1.predict(queryBERT1['input_ids'])
maxProb = np.argmax(BERTPred['issue'], axis=1) 
print('Sentiment Prediction: ',predLabel[maxProb[0]])
```
